```python
# function to parse data from pdf file and return in text format
def parse_pdf(file_path) -> str:
  txt = ""
  doc = fitz.open(file_path)
  for page in doc:
    txt += page.get_text("text") + "\n"
  return txt
```

```python
data = parse_pdf(file_path)
```
